```python
class PromptLibrary:

    def __init__(self):
        self.templates = {
            "basic": PromptTemplate(
                input_variables=["topic"],
                template="Explain {topic} in simple terms for beginners"
            ),

            "multi": PromptTemplate(
                input_variables=["topic", "audience", "tone"],
                template="Explain {topic} for {audience} in a {tone} tone"
            ),

            "teaching": PromptTemplate(
                input_variables=["topic"],
                template="Break down {topic} into clear steps with examples"
            ),

            "interview": PromptTemplate(
                input_variables=["topic"],
                template="Generate 3 challenging interview questions on {topic}"
            ),

            "story": PromptTemplate(
                input_variables=["topic"],
                template="Explain {topic} through a real-life story"
            )
        }

    def get(self, name):
        return self.templates.get(name)
```

```python
class InputValidator:

    def __init__(self):
        self.valid_audience = ["beginner", "intermediate", "expert"]
        self.valid_tone = ["formal", "casual", "fun"]

    def validate(self, audience, tone):
        audience = audience.lower()
        tone = tone.lower()

        if audience not in self.valid_audience:
            audience = "beginner"

        if tone not in self.valid_tone:
            tone = "formal"

        return audience, tone
```

```python
class PromptEngine:

    def __init__(self):
        self.library = PromptLibrary()
        self.validator = InputValidator()

    def generate(self, topic, audience, tone, style):

        audience, tone = self.validator.validate(audience, tone)

        if style in ["teaching", "interview", "story"]:
            template = self.library.get(style)
            return template.format(topic=topic)

        template = PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Explain {topic} for {audience} in a {tone} tone"
        )

        return template.format(topic=topic, audience=audience, tone=tone)
```

```python
class ChatEngine:

    def __init__(self):
        self.roles = {
            "teacher": "You explain concepts clearly step by step",
            "interviewer": "You ask difficult and analytical questions",
            "motivator": "You inspire and simplify learning"
        }

    def build(self, role, topic):

        system_msg = self.roles.get(role, "You are a helpful assistant")

        chat = ChatPromptTemplate.from_messages([
            ("system", system_msg),
            ("human", "Explain {topic}")
        ])

        return chat.format_messages(topic=topic)
```

```python
def variation_engine(topic):

    templates = [
        "Explain {topic} in one paragraph",
        "Summarize {topic} in bullet points",
        "Explain {topic} with a real-world example",
        "Explain {topic} in a humorous way"
    ]

    results = []

    for temp in templates:
        t = PromptTemplate(input_variables=["topic"], template=temp)
        results.append(t.format(topic=topic))

    return results

variation_engine("Machine Learning")
```

```python
engine = PromptEngine()

test_cases = [
    ("AI", "beginner", "fun", "basic"),
    ("Python", "expert", "formal", "multi"),
    ("ML", "intermediate", "casual", "teaching"),
    ("Data Science", "beginner", "formal", "story"),
    ("Deep Learning", "expert", "fun", "interview")
]

for t in test_cases:
    print(engine.generate(*t))
```

```python
def pipeline(topic, audience, tone, style, role):

    engine = PromptEngine()
    chat = ChatEngine()

    prompt = engine.generate(topic, audience, tone, style)
    chat_msg = chat.build(role, topic)

    return {
        "generated_prompt": prompt,
        "chat_prompt": chat_msg
    }

pipeline("Neural Networks", "beginner", "fun", "story", "teacher")
```
