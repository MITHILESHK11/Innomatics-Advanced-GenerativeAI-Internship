```python
def preprocess_text(text):
    try:
        if not isinstance(text, str) or text.strip() == "":
            return [], ""
        text = re.sub(r"http\S+|www\S+|https\S+", "", text)
        text = re.sub(r"\S+@\S+", "", text)
        text = re.sub(r"\d+", "", text)
        text = text.lower()
        text = re.sub(r'(.)\1{2,}', r'\1\1', text)
        text = text.translate(str.maketrans('', '', string.punctuation))
        text = re.sub(r"\s+", " ", text).strip()
        tokens = text.split()
        tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]
        cleaned_sentence = " ".join(tokens)
        return tokens, cleaned_sentence
    except Exception as e:
        return [], ""
```

```python
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

results = []

for sentence in test_sentences:
    tokens, cleaned = preprocess_text(sentence)
    
    results.append({
        "original": sentence,
        "tokens": tokens,
        "cleaned": cleaned
    })

# Display
for r in results:
    print("\nOriginal:", r["original"])
    print("Tokens:", r["tokens"])
    print("Cleaned:", r["cleaned"])
```

```python
analytics = []
for r in results:
    tokens = r["tokens"]
    if len(tokens) == 0:
        total = unique = avg_len = 0
    else:
        total = len(tokens)
        unique = len(set(tokens))
        avg_len = sum(len(t) for t in tokens) / total
    analytics.append({
        "sentence": r["original"],
        "total_tokens": total,
        "unique_tokens": unique,
        "avg_token_length": round(avg_len, 2)
    })
for a in analytics:
    print("\nSentence:", a["sentence"])
    print("Total Tokens:", a["total_tokens"])
    print("Unique Tokens:", a["unique_tokens"])
    print("Avg Length:", a["avg_token_length"])
```

```python
all_tokens = []

for r in results:
    all_tokens.extend(r["tokens"])

counter = Counter(all_tokens)

print("Top 10 Frequent Words:", counter.most_common(10))
print("Top 5 Least Frequent Words:", counter.most_common()[-5:])
```

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, cleaned = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(cleaned)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

pipeline_output = full_pipeline(test_sentences)
pipeline_output
```
